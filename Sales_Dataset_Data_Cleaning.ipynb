```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Messy_data.csv')
```

```python

text_columns = ['CustomerName', 'City', 'Product', 'OrderStatus']

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df['City'] = df['City'].str.title()
```

```python
df['OrderStatus'] = df['OrderStatus'].str.title()
```

```python
df[['CustomerName', 'City', 'OrderStatus']].head(10)
```

```python
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')

df['UnitPrice'] = pd.to_numeric(
    df['UnitPrice'].str.replace(',', '', regex=False),
    errors='coerce'
)

df['CustomerAge'] = pd.to_numeric(
    df['CustomerAge'],
    errors='coerce'
)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['CustomerName'] = df['CustomerName'].fillna('Unknown')

df['City'] = df['City'].fillna('Unknown')
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
```

```python
df['UnitPrice'] = df['UnitPrice'].fillna(df['UnitPrice'].median())
```

```python
df['CustomerAge'] = df['CustomerAge'].fillna(df['CustomerAge'].median())
```

```python
df['OrderStatus'] = df['OrderStatus'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['OrderDate'] = pd.to_datetime(
    df['OrderDate'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
```

```python
df['OrderDate'].head(10)
```

```python
df = df.dropna(subset=['OrderDate'])
```

```python
df['OrderDate'].isnull().sum()
```

```python
df.loc[
    (df['CustomerAge'] < 18) | (df['CustomerAge'] > 100),
    'CustomerAge'
] = np.nan
```

```python
df['CustomerAge'] = df['CustomerAge'].fillna(
    df['CustomerAge'].median()
)
```

```python
Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

df.loc[
    (df['Quantity'] <= 0) | (df['Quantity'] > upper_limit),
    'Quantity'
] = np.nan
```

```python
df['Quantity'] = df['Quantity'].fillna(
    df['Quantity'].median()
)
```

```python
Q1 = df['UnitPrice'].quantile(0.25)
Q3 = df['UnitPrice'].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

df.loc[
    df['UnitPrice'] > upper_limit,
    'UnitPrice'
] = np.nan
```

```python
df['UnitPrice'] = df['UnitPrice'].fillna(
    df['UnitPrice'].median()
)
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']
```

```python
df.to_csv('cleaned_sales_dataset.csv', index=False)
```
